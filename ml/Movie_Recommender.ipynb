```python
import numpy as np
import pandas as pd
```

```python
movie_data = pd.read_csv('artifacts/tmdb_5000_movies.csv')
```

```python
creadis_data = pd.read_csv('artifacts/tmdb_5000_credits.csv')
```

```python
movie_data.head()
```

```python
creadis_data.head()
```

```python
df = movie_data.merge(creadis_data,on='title')
```

```python
df.head()
```

```python
df.info()
```

```python
# id 
# title
# genres
# overview
# keywords
# cast
# crew
df = df[['id','title','overview','keywords','genres','cast','crew']]
```

```python
df.head()
```

```python
import ast
def covert(values):
    l = []
    for i in ast.literal_eval(values):
        l.append(i['name'])
    return l
```

```python
df['genres'] = df['genres'].apply(covert)
```

```python
df['keywords'] = df['keywords'].apply(covert)
```

```python
df.head()
```

```python
import ast
def covert_cast(values):
    l = []
    count = 0
    for i in ast.literal_eval(values):
        if count != 3:
            l.append(i['name'])
            count+=1
        else:
            break
    return l
```

```python
df['cast'] = df['cast'].apply(covert_cast)
```

```python
df.head()
```

```python
import ast
def covert_crew(values):
    l = []
    for i in ast.literal_eval(values):
        if i['job'] == 'Director':
            l.append(i['name'])
            break
    return l
```

```python
df['crew'] = df['crew'].apply(covert_crew)
```

```python
df.info()
```

```python
df.head()
```

```python
import re

# Preprocess the 'overview' column
def preprocess_overview(overview):
    if pd.isna(overview):
        return []
    # Remove punctuation and convert to lowercase
    overview = re.sub(r'[^\w\s]', '', overview.lower())
    return overview.split()

df['overview'] = df['overview'].apply(preprocess_overview)
```

```python
df.head()
```

```python
df['overview'] = df['overview'].apply(lambda x:[i.replace(' ','') for i in x])
df['keywords'] = df['keywords'].apply(lambda x:[i.replace(' ','') for i in x])
df['genres'] = df['genres'].apply(lambda x:[i.replace(' ','') for i in x])
df['cast'] = df['cast'].apply(lambda x:[i.replace(' ','') for i in x])
df['crew'] = df['crew'].apply(lambda x:[i.replace(' ','') for i in x])
```

```python
df['tags'] = df['overview'] + df['keywords'] + df['genres'] + df['cast'] + df['crew']
df.head()
```

```python
df['tags'] = df['tags'].apply(lambda x: ' '.join(x))
df.head()
```

```python
df['tags'] = df['tags'].apply(lambda x: x.lower())
df.head()
```

```python
df = df[['id','title','tags']]
df.head()
```

```python
df.to_csv('artifacts/movies_data.csv')
```
